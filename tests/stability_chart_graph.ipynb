```python
import SADDlE
import numpy as np
import matplotlib.pyplot as plt
import tikzplotlib
from matplotlib.lines import Line2D
```

```python
def dde (a,b):
    return SADDlE.generalDDE(1, [1], None, None , np.array([[a]]), [np.array([[b]])], None, None, False)
```

```python
a = np.linspace(-20, 20, 100)
b = np.linspace(-20, 20, 100)
result =SADDlE.generateStabilityTable(dde, (a,b), 30, False)
```

```python
from matplotlib import colormaps
cmap = colormaps['viridis']
plt.imshow(result, cmap=cmap, extent = [-20, 20, 20, -20])
plt.xlabel("b")
plt.ylabel("a")
custom_lines = [Line2D([0], [0], color=cmap(0.), lw=4),
                Line2D([0], [0], color=cmap(1.), lw=4)]
plt.legend(custom_lines, ['not stable', 'stable'])
tikzplotlib.save("stability_chart_graph.tex")
```
